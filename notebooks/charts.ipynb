# hzero charts

Builds the hzero charts in the house style (see `DESIGN.md` in the repository root) and exports them for the website and for slides.

* **Website:** `web/public/charts/<name>.svg`, cropped to the chart, used as `<img src="charts/<name>.svg" alt="…">`
* **Slides:** `exports/slides/<name>.png` (3840 × 2160) and `exports/slides/<name>.svg` (1920 × 1080 frame)

To make a new chart: load your data into a DataFrame, call `bar_chart` or `bar_line_chart`, then `export`.

> The numbers below are **read back from the two reference visuals** and are placeholders. Replace them with the real data.

```python
# Parameters (papermill can override these)
web_dir = "../web/public/charts"
slides_dir = "../exports/slides"
```

```python
import pandas as pd
import json
from hzero_charts import bar_chart, bar_line_chart, series_chart, theme, export
```

## Spending and systems per year

Stacked bars on the left axis (MEUR), two lines on the right axis (number of systems). The series names are placeholders: the reference had no legend.

```python
per_year = pd.DataFrame({
    "year":            [2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023, 2024, 2025],
    "spending_base":   [28.0, 28.0, 28.0, 42.1, 42.1, 28.0, 42.1, 84.1, 112.1, 130.8],   # sage, bottom of the stack
    "spending_top":    [28.0, 0.0, 42.1, 28.0, 28.0, 56.1, 42.1, 224.3, 299.1, 373.8],   # dark green, top of the stack
    "systems_line_1":  [5.6, 4.2, 10.3, 15.0, 20.6, 27.6, 34.6, 69.2, 111.2, 145.3],   # sand line
    "systems_line_2":  [1.9, 2.8, 5.1, 7.9, 11.2, 14.5, 19.6, 27.1, 39.3, 57.9],       # coral line
})

fig = bar_line_chart(
    categories=per_year["year"].astype(str),
    bars={"Spending, base": per_year["spending_base"], "Spending, top": per_year["spending_top"]},
    lines={"Systems, line 1": per_year["systems_line_1"], "Systems, line 2": per_year["systems_line_2"]},
    left_ticks=[500, 1000, 1500], right_ticks=[50, 100, 150],
    left_title="Spending (MEUR)", right_title="Number of systems",
)
export(fig, "spending-and-systems", web_dir=web_dir, slides_dir=slides_dir)
```

## Systems by segment

```python
segments = pd.DataFrame({
    "segment": ["R&D\nFederal", "HPC\nData center", "R&D\nAcademia", "Defense\nMilitary", "R&D\nCommercial", "Industrial\nEnterprise"],
    "systems": [20, 18.7, 15.2, 4, 3.1, 3.1],
})

fig = bar_chart(segments["segment"], segments["systems"], ticks=[4, 8, 12, 16, 20], title="Number of systems")
export(fig, "systems-by-segment", web_dir=web_dir, slides_dir=slides_dir)
```

## Dark slides

The same charts in the dark slide style of `28.svg`: page `#091314`, plot backdrop `#091314` to `#102223`, gridlines `#3A4243`, text `#F7F7F7`, bars mint and hzero cyan, lines green and coral. `with theme("dark")` switches the palette; sizes and positions stay identical to the light charts.

```python
with theme("dark"):
    fig = bar_line_chart(
        categories=per_year["year"].astype(str),
        bars={"Spending, base": per_year["spending_base"], "Spending, top": per_year["spending_top"]},
        lines={"Systems, line 1": per_year["systems_line_1"], "Systems, line 2": per_year["systems_line_2"]},
        left_ticks=[500, 1000, 1500], right_ticks=[50, 100, 150],
        left_title="Spending (MEUR)", right_title="Number of systems",
    )
    export(fig, "spending-and-systems-dark", web_dir=web_dir, slides_dir=slides_dir)
```

## Estimated success probability (dark slide)

hzero's simulator over about 3,600 time units: the estimated success probability of the running workload, with the two calibration windows tinted and dashed averages (95.68% while running, the calibration averages in mint).

The values in `data/success_probability.json` were read off the simulator's screenshot of 11 November 2026, so they are an approximation. Replace the file with the simulator's own export when it is available; nothing else needs to change.

```python
sp = json.load(open("data/success_probability.json"))
xs, ys = zip(*sp["points"])
runs = [b for b in sp["bands"] if b[0] == "run"]

with theme("dark"):
    fig = series_chart(
        xs, ys, x_ticks=[4000, 5000, 6000], y_ticks=[90, 95, 100], y_bottom=88.4,
        x_range=tuple(sp["x_range"]),
        bands=[tuple(b) for b in sp["bands"]],
        averages=[("run", r[1], r[2], sp["avg_run"]) for r in runs[:2]]
                 + [("calibrate", a0, a1, v) for a0, a1, v in sp["avg_calibrate"]],
        label=(f"Avg {sp['avg_run']:.2f}%", 3580, sp["avg_run"]),
        y_title="Estimated success probability (%)", x_title="Simulated time units",
    )
    export(fig, "success-probability", web_dir=web_dir, slides_dir=slides_dir)
```
